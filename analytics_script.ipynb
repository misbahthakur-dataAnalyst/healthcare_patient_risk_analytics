```python
import pandas as pd
import numpy as np

data_clean= {
    "patient_ID":[101,102,103,104,105],
    "patient_name":["misbah","arjun","neha","rahul"," madeeha"],
    "age":[95,np.nan,31,45,62],
    "clinical_dept":["biotech lab","zoology reseach","biochem clinic",np.nan,"biotech lab"],
    "glucose_level":[95,140,np.nan,110,180],
    "blood_pressure":[120,135,130,140,np.nan]
}

# converting into a standard dataframe variable
df= pd.DataFrame(data_clean)
df.to_csv("hospital_input.csv",index=False)

print("___ mubarak ho! hamara project data matrix ready hai___")
print(df)
```

```python
import pandas as pd

# 1. active file ko refresh variable mai load kijiyai
df_clean= pd.read_csv("hospital_input.csv")
df_clean["clinical_dept"]= df_clean["clinical_dept"].fillna("general OPD")
df_clean["age"]= df_clean["age"].fillna(35)
df_clean= df_clean.bfill().ffill()

print("___ cleaned medical informatics table___")
print(df_clean)
```

```python
df_clean["risk_level"]= df_clean["glucose_level"].apply(lambda x:"high risk"if x >=140 else "normal")

print("___ premium patient risk analysis dashboard ___")
print(df_clean)

df_clean.to_csv("cleaned_patient_analytics.csv",index=False)
print("\n mubarak ho! final clean file 'cleaned_patient_analytics.csv' save ho chuki hai.")
```
